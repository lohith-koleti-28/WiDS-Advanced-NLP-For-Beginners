```python
from sklearn.feature_extraction.text import TfidfVectorizer

v = TfidfVectorizer()
```

```python
corpus = [
     "Thor eating pizza, Loki is eating pizza, Ironman ate pizza already",
    "Apple is announcing new iphone tomorrow",
    "Tesla is announcing new model-3 tomorrow",
    "Google is announcing new pixel-6 tomorrow",
    "Microsoft is announcing new surface tomorrow",
    "Amazon is announcing new eco-dot tomorrow",
    "I am eating biryani and you are eating grapes"
]
```

```python
v.fit(corpus)
v.get_feature_names_out()
```

```python
v.vocabulary_
```

```python
all_feature_names = v.get_feature_names_out()
idf_scores = {}
for word in all_feature_names:
    i = v.vocabulary_.get(word)
    idf_scores[word] = v.idf_[i]

print(idf_scores.items())
```

```python
idf_scores = dict(sorted(idf_scores.items(),key=lambda x:x[1]))
idf_scores
```

```python
import pandas as pd

df = pd.read_csv('Data Files/Ecommerce_data.csv')

print(df.info(),'\n\n',df.describe())

print(df.head())
```

```python
df['label'].value_counts()
```

```python
df['target'] = df['label'].map({
    'Household' : 0,
    'Electronics' : 1,
    'Clothing & Accessories' : 2,
    'Books' : 3
})

df.head()
```

```python
df['target'].value_counts()
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(df['Text'],df['target'], test_size=0.2,random_state=2024,stratify=df['target'])
print(X_test.shape,y_train.shape)
```

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import classification_report
from sklearn.pipeline import Pipeline
from sklearn.ensemble import RandomForestClassifier
```

```python
p1 = Pipeline([
    ('vectorizer',TfidfVectorizer()),
    ('knn',KNeighborsClassifier())
])

p2 = Pipeline([
    ('vectorizer',TfidfVectorizer()),
    ('rf',RandomForestClassifier(n_estimators=15))
])

p3 = Pipeline([
    ('vectorizer',TfidfVectorizer()),
    ('nb',MultinomialNB())
])

p1.fit(X_train,y_train)
p2.fit(X_train,y_train)
p3.fit(X_train,y_train)

print('Classification For KNN\n')
print(classification_report(y_test,p1.predict(X_test)))

print('Classification For RF\n')
print(classification_report(y_test,p2.predict(X_test)))

print('Classification For NB\n')
print(classification_report(y_test,p3.predict(X_test)))
```

```python
import spacy

nlp = spacy.load('en_core_web_sm')

def preprocess(text):
    filtered_tokens = []
    doc = nlp(text)

    for token in doc:
        if token.is_stop or token.is_punct:
            continue
        filtered_tokens.append(token.lemma_)
    return " ".join(filtered_tokens)
```

```python
df['preprocessed'] = df['Text'].apply(preprocess)
```

```python
df.head()
```

```python
X_train, X_test, y_train, y_test = train_test_split(df['preprocessed'],df['target'], test_size=0.2,random_state=2024,stratify=df['target'])
print(X_test.shape,y_train.shape)
```

```python
p1.fit(X_train,y_train)
p2.fit(X_train,y_train)
p3.fit(X_train,y_train)

print('Classification For KNN\n')
print(classification_report(y_test,p1.predict(X_test)))

print('Classification For RF\n')
print(classification_report(y_test,p2.predict(X_test)))

print('Classification For NB\n')
print(classification_report(y_test,p3.predict(X_test)))
```

### Modelling without Preprocessing

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(df['Comment'],df['target'], test_size=0.2,random_state=2024,stratify=df['target'])
```

```python
print(X_train.shape,X_test.shape)
```

```python
from sklearn.feature_extraction.text import CountVectorizer
```

### Attempt 1
- CountVectorizer with only trigrams
- RF as classifier

```python
from sklearn.pipeline import Pipeline

p1 = Pipeline([
    ('vectorizer',CountVectorizer(ngram_range=(3,3))),
    ('rf',RandomForestClassifier(n_estimators=20))
])

p1.fit(X_train,y_train)

print(classification_report(y_test,p1.predict(X_test)))
```

### Attempt 2
- CV for uni and bigrams
- NB as classifier

```python
p2 = Pipeline([
    ('vectorizer',CountVectorizer(ngram_range=(1,2))),
    ('rf',MultinomialNB())
])

p2.fit(X_train,y_train)

print(classification_report(y_test,p2.predict(X_test)))
```

### Attempt 3

- RF as classifier

```python
p3 = Pipeline([
    ('vectorizer',CountVectorizer(ngram_range=(1,2))),
    ('rf',RandomForestClassifier(n_estimators=20))
])

p3.fit(X_train,y_train)

print(classification_report(y_test,p3.predict(X_test)))
```

### Attempt 4
- TF-IDF
- RF
