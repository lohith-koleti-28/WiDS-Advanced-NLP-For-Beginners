```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('spam.csv')
print(df.info(),'\n',df.describe())
```

```python
df['spam'] = df['Category'].apply(lambda x: 1 if x == 'spam' else 0)
df
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(df['Message'],df['spam'],test_size=0.2)
```

```python
from sklearn.feature_extraction.text import CountVectorizer

v = CountVectorizer()

X_train_cv = v.fit_transform(X_train.values)
X_train_cv
```

```python
X_train_cv.shape
```

```python
v.get_feature_names_out()[1000:1050]
```

```python
print(v.get_feature_names_out()[6867])
v.vocabulary_

```

```python
from sklearn.naive_bayes import MultinomialNB

model = MultinomialNB()
model.fit(X_train_cv,y_train)
```

```python
X_test_cv = v.transform(X_test)
y_pred = model.predict(X_test_cv)
```

```python
from sklearn.metrics import classification_report, confusion_matrix

rpt = classification_report(y_test,y_pred)
cm = confusion_matrix(y_test,y_pred)
print(cm,'\n\n',rpt)
```

```python
from sklearn.pipeline import Pipeline

clf = Pipeline([
    ('vectoriser',CountVectorizer()),
    ('nb', MultinomialNB())
])

clf.fit(X_train,y_train)
```

```python
clf.score(X_test,y_test)
```

```python
i = 2
print("Message",df['Message'][i])
print("Prediction",'spam' if clf.predict([df['Message'][i]]).item() == 1 else 'ham')
```

```python
#Import necessary libraries

import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.ensemble import RandomForestClassifier
from  sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import MultinomialNB
from sklearn.pipeline import Pipeline
from sklearn.metrics import classification_report
```

```python
df = pd.read_csv('movies_sentiment_data.csv')
print(df.info(),'\n\n',df.describe())
```

```python
print(df.shape, df.head())
```

```python
df['Category'] = df['sentiment'].apply(lambda x : 1 if x == 'positive' else 0)
df.head()
```

```python
df['Category'].value_counts()
```

```python
X_train, X_test, y_train, y_test = train_test_split(df['review'],df['sentiment'],test_size=0.2)
```

```python
pipeline = Pipeline([
    ('vectoriser',CountVectorizer()),
    ('rf',RandomForestClassifier(n_estimators=50, criterion='entropy'))
])

pipeline.fit(X_train,y_train)
```

```python
y_pred = pipeline.predict(X_test)
print(classification_report(y_test,y_pred))
```

```python
pipeline_2 = Pipeline([
    ('vectoriser',CountVectorizer()),
    ('knn',KNeighborsClassifier(n_neighbors=10, metric='euclidean'))
])

pipeline_3 = Pipeline([
    ('vectoriser',CountVectorizer()),
    ('nb',MultinomialNB())
])

pipeline_2.fit(X_train,y_train)
pipeline_3.fit(X_train,y_train)

y_pred2 = pipeline_2.predict(X_test)
y_pred3 = pipeline_3.predict(X_test)

print("Classification Report for KNN", classification_report(y_test,y_pred2))
print("Classification Report for NB", classification_report(y_test,y_pred3))
```
